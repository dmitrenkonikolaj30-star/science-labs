```python
print("4.1 Реалізувати процедуру для формування вибірки ряду VHI за областю та роком")
```

```python
def get_vhi_series(df, province_id, year):
    result = df[
        (df["Area_ID"] == province_id) &
        (df["Year"] == year)
    ][["Week", "VHI"]]

    return result
```

```python
print("4.2 Максимальне та мінімальне значення VHI")
```

```python
def get_vhi_extremes(df, province_id, year):
    data = df[
        (df["Area_ID"] == province_id) &
        (df["Year"] == year)
    ]

    return {
        "min_vhi": data["VHI"].min(),
        "max_vhi": data["VHI"].max()
    }
```

```python
print("4.3 Роки з екстремальними значеннями VHI")
```

```python
def get_extreme_years(df, province_id):
    data = df[df["Area_ID"] == province_id]

    min_row = data.loc[data["VHI"].idxmin()]
    max_row = data.loc[data["VHI"].idxmax()]

    return {
        "year_min": min_row["Year"],
        "vhi_min": min_row["VHI"],
        "year_max": max_row["Year"],
        "vhi_max": max_row["VHI"]
    }
```

```python
get_extreme_years(df, 1)
```

```python
print("4.4 Пошук років з екстремальною посухою")
```

```python
def drought_years(df, province_id, threshold=15):
    data = df[
        (df["Area_ID"] == province_id) &
        (df["VHI"] < threshold)
    ]

    return sorted(data["Year"].unique())
```

```python
drought_years(df, 1)
```

```python
print("4.5 Пошук років з помірною посухою")
```

```python
def moderate_drought_years(df, province_id):
    data = df[
        (df["Area_ID"] == province_id) &
        (df["VHI"] >= 15) &
        (df["VHI"] < 35)
    ]

    return sorted(data["Year"].unique())
```

```python
moderate_drought_years(df, 1)
```

```python
print("4.6 Пошук років з сильною посухою")
```

```python
def severe_drought_years(df, province_id):
    data = df[
        (df["Area_ID"] == province_id) &
        (df["VHI"] >= 0) &
        (df["VHI"] < 15)
    ]

    return sorted(data["Year"].unique())
```

```python
severe_drought_years(df, 1)
```
