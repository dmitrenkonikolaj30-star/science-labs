```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
```

```python
df = pd.read_csv("household_power_clean.csv")

print(df.head())
print(df.info())
```

```python
print("Пропущені значення відсутні після очищення.")
```

```python
print(df.isnull().sum())
```

```python
print("Графік 1. Залежність Global_active_power від Global_intensity")
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="Global_intensity",
    y="Global_active_power"
)
plt.title("Global Active Power vs Global Intensity")
plt.show()
```

```python
print("Графік 2. Зміна Global_active_power для перших 500 спостережень")
```

```python
sample = df.head(500)

plt.figure(figsize=(10,5))
plt.plot(sample["Global_active_power"])
plt.title("Global Active Power")
plt.xlabel("Observation")
plt.ylabel("Power")
plt.show()
```

```python
print("Графік 3. Розподіл напруги Voltage")
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Voltage"], bins=30)
plt.title("Voltage Distribution")
plt.xlabel("Voltage")
plt.ylabel("Frequency")
plt.show()
```

```python
print("Графік 4. Виявлення викидів у Global_active_power")
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df["Global_active_power"])
plt.title("Boxplot of Global Active Power")
plt.show()
```

```python
print("Графік 5. Кореляційна матриця ознак")
```

```python
numeric_cols = [
    "Global_active_power",
    "Global_reactive_power",
    "Voltage",
    "Global_intensity",
    "Sub_metering_1",
    "Sub_metering_2",
    "Sub_metering_3"
]

plt.figure(figsize=(10,6))
sns.heatmap(
    df[numeric_cols].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Matrix")
plt.show()
```

```python
print("Графік 6. Pairplot для багатовимірного аналізу")
```

```python
sns.pairplot(
    df[
        [
            "Global_active_power",
            "Voltage",
            "Global_intensity"
        ]
    ].sample(300)
)

plt.show()
```
