```python
print("Частина 2. Аналіз Individual Household Electric Power Consumption Dataset")
```

```python
import pandas as pd
import timeit

df = pd.read_csv(
    "data_power/household_power_consumption.txt",
    sep=";",
    low_memory=False
)

print("Перші 5 рядків датасету:")
display(df.head())

print("Розмір датасету:", df.shape)
```

```python
print("Кількість пропусків до очищення:")
print(df.isnull().sum())

numeric_cols = [
    "Global_active_power",
    "Global_reactive_power",
    "Voltage",
    "Global_intensity",
    "Sub_metering_1",
    "Sub_metering_2",
    "Sub_metering_3"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df_clean = df.dropna()

print("Розмір датасету після очищення:", df_clean.shape)

print("Кількість пропусків після очищення:")
print(df_clean.isnull().sum())

df_clean.to_csv("household_power_clean.csv", index=False)
print("Очищений датасет збережено у файл household_power_clean.csv")
```

```python
print("Завдання 1. Обрати всі записи, у яких Global_active_power > 5 кВт")

def power_over_5kw():
    return df_clean[df_clean["Global_active_power"] > 5]

result_1 = power_over_5kw()

print("Кількість записів:", len(result_1))
display(result_1.head(20))
```

```python
print("Завдання 2. Обрати записи, де струм 19-20 А, а Sub_metering_2 > Sub_metering_1 + Sub_metering_3")

def current_19_20():
    data = df_clean[
        (df_clean["Global_intensity"] >= 19) &
        (df_clean["Global_intensity"] <= 20)
    ]
    
    result = data[
        data["Sub_metering_2"] > 
        (data["Sub_metering_1"] + data["Sub_metering_3"])
    ]
    
    return result

result_2 = current_19_20()

print("Кількість записів:", len(result_2))
display(result_2.head(20))
```

```python
print("Завдання 3. Обрати випадково 500000 записів без повторів і знайти середні значення трьох груп споживання")

def random_500k():
    return df_clean.sample(
        n=500000,
        replace=False,
        random_state=42
    )

sample_500k = random_500k()

means = sample_500k[
    ["Sub_metering_1", "Sub_metering_2", "Sub_metering_3"]
].mean()

print("Середні значення:")
print(means)
```

```python
print("Завдання 4. Після 18:00, понад 6 кВт, основне споживання припадає на групу 2")

def evening_consumption():
    data = df_clean[
        (df_clean["Time"] > "18:00:00") &
        (df_clean["Global_active_power"] > 6)
    ]
    
    data = data[
        (data["Sub_metering_2"] > data["Sub_metering_1"]) &
        (data["Sub_metering_2"] > data["Sub_metering_3"])
    ]
    
    half = len(data) // 2
    
    first_half = data.iloc[:half]
    second_half = data.iloc[half:]
    
    result_first = first_half.iloc[::3]
    result_second = second_half.iloc[::4]
    
    return pd.concat([result_first, result_second])

result_4 = evening_consumption()

print("Кількість записів:", len(result_4))
display(result_4.head(20))
```

```python
print("Профілювання часу виконання функцій за допомогою timeit")

t1 = timeit.timeit(lambda: power_over_5kw(), number=100)
t2 = timeit.timeit(lambda: current_19_20(), number=100)
t3 = timeit.timeit(lambda: random_500k(), number=10)
t4 = timeit.timeit(lambda: evening_consumption(), number=100)

print("Вибірка 1:")
print("Час за 100 запусків:", t1)
print("Середній час:", t1 / 100)

print("\nВибірка 2:")
print("Час за 100 запусків:", t2)
print("Середній час:", t2 / 100)

print("\nВибірка 3:")
print("Час за 10 запусків:", t3)
print("Середній час:", t3 / 10)

print("\nВибірка 4:")
print("Час за 100 запусків:", t4)
print("Середній час:", t4 / 100)
```

```python
print("Нормалізація вибраного датасету")

numeric_cols = [
    "Global_active_power",
    "Global_reactive_power",
    "Voltage",
    "Global_intensity",
    "Sub_metering_1",
    "Sub_metering_2",
    "Sub_metering_3"
]

df_normalized = df_clean.copy()

for col in numeric_cols:
    df_normalized[col] = (
        (df_clean[col] - df_clean[col].min()) /
        (df_clean[col].max() - df_clean[col].min())
    )

display(df_normalized[numeric_cols].head())
```

```python
print("Стандартизація вибраного датасету")

df_standardized = df_clean.copy()

for col in numeric_cols:
    df_standardized[col] = (
        (df_clean[col] - df_clean[col].mean()) /
        df_clean[col].std()
    )

display(df_standardized[numeric_cols].head())
```

```python
print("Коефіцієнт Пірсона та Спірмена для Global_active_power і Global_intensity")

pearson = df_clean["Global_active_power"].corr(
    df_clean["Global_intensity"],
    method="pearson"
)

spearman = df_clean["Global_active_power"].corr(
    df_clean["Global_intensity"],
    method="spearman"
)

print("Коефіцієнт Пірсона:", pearson)
print("Коефіцієнт Спірмена:", spearman)
```

```python
print("One Hot Encoding категоріального атрибута Date")

sample_dates = df_clean["Date"].drop_duplicates().head(5)

df_ohe = df_clean[
    df_clean["Date"].isin(sample_dates)
].copy()

df_encoded = pd.get_dummies(
    df_ohe,
    columns=["Date"]
)

print("Розмір після One Hot Encoding:", df_encoded.shape)
display(df_encoded.head())

print("Стовпці після One Hot Encoding:")
print(df_encoded.columns)
```
